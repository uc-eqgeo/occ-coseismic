```python
import geopandas as gpd
from shapely.geometry import Point
import numpy as np

"""
Take geojson of sites (e.g validation sites) and round their locations to the semi-regular grid for occ
"""

rounding = 1
sites = gpd.read_file('../sites/validation_sites_KC.geojson')
name_col = 'Site'

original_location = sites.geometry
sites['geometry'] = sites['geometry'].apply(lambda x: Point(np.round(x.x / rounding) * rounding, np.round(x.y / rounding) * rounding))
sites['rounding_distance'] = sites.geometry.distance(original_location)
if name_col in sites.columns:
    sites = sites.rename(columns={name_col: 'siteId'})
else:
    sites['siteId'] = sites.geometry.apply(lambda x: f"{x.x:.0f}_{x.y:.0f}")
sites['Lon'], sites['Lat'] = sites.geometry.x, sites.geometry.y
sites['Height'] = 0

sites.to_file('../sites/validation_sites_KC_1m.geojson', driver="GeoJSON")
```
